```python
from transformers import AutoTokenizer, AutoModelForCausalLM
import torch
```

```python
# 1. Load the model and tokenizer

model_name = "gpt2"

tokenizer = AutoTokenizer.from_pretrained(model_name)

model = AutoModelForCausalLM.from_pretrained(model_name)
```

```python
#2 Prepare the input text

text = "The weather looks nice"

# Convert text into tokens

input = tokenizer(text, return_tensors="pt")

print("Tokens:", tokenizer.convert_ids_to_tokens(input["input_ids"][0]))

```

```python
#3 to run the model
# no_grad means no gradients to be calculated , also we dont want to train the model just want forward pass
with torch.no_grad():
  outputs = model(**input)

#Get logits(raw scores) for the next token
#[0, -1] means 1 st batch , last position

next_token_logits = outputs.logits[0, -1]

#top 5 most likely tokens
top5 = torch.topk(next_token_logits, 5)
```

```python
#4 show the results
print("\nTop 5 predicted next tokens:")
for score, idx in zip(top5.values, top5.indices):
  token = tokenizer.decode([idx])
  print(f" {token:15s} score: {score.item():.2f}")
```
